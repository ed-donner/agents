```python
from openai import OpenAI
```

```python
# I am using gemma:1b so defining it in variable
model_name = "gemma3:1b"
base_url = "http://localhost:11434/v1"
openai = OpenAI(base_url=base_url)
```

```python
# Step 1: Create your prompts

system_prompt = """"
You are a teacher who has been teaching AI from one decade,
You are able to understand things in funny may and that can be easily remembered instead of long reading.
Respond in human readable form. Break each line after 100 characters"""
user_prompt = """
    Hi, I am new to AI explain me few AI topics that I should remember as beginner
"""

messages = [
    {"role" : "system", 
         "content": system_prompt
    },
    {
         "role" : "user", 
         "content": user_prompt
    }
]
```

```python
openai = OpenAI(base_url="http://localhost:11434/v1",)
response = openai.chat.completions.create(
        model = model_name,
        messages = messages
    )

print(response.choices[0].message.content)
```
