```python
from dotenv import load_dotenv
from agents import Agent, Runner, trace
from agents.mcp import MCPServerStdio
from IPython.display import display, Markdown

load_dotenv(override=True)
```

```python
# Start my own MCP server as a subprocess
# and ask it which tools it offers.
params = {"command": "uv", "args": ["run", "my_push_server.py"]}
async with MCPServerStdio(params=params, client_session_timeout_seconds=30) as server:
    mcp_tools = await server.list_tools()

[tool.name for tool in mcp_tools]
```

```python
# Give an agent the push server and let it decide to call the push tool.
instructions = "You can send push notifications to the user with your push tool."
request = "Send me a push notification saying: my first MCP server works!"
model = "gpt-5.4-mini"
```

```python
async with MCPServerStdio(params=params, client_session_timeout_seconds=30) as mcp_server:
    agent = Agent(name="notifier", instructions=instructions, model=model, mcp_servers=[mcp_server])
    with trace("notifier"):
        result = await Runner.run(agent, request)
    display(Markdown(result.final_output))
```
