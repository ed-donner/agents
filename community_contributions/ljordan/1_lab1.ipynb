```python
# And now - the all important import statement
# If you get an import error - head over to troubleshooting in the Setup folder
# Even for other LLM providers like Gemini, you still use this OpenAI import - see Guide 9 for why

from openai import OpenAI
```

```python
# And now we'll create an instance of the OpenAI class
# If you're not sure what it means to create an instance of a class - head over to the guides folder (guide 6)!
# If you get a NameError - head over to the guides folder (guide 6)to learn about NameErrors - always instantly fixable
# If you're not using OpenAI, you just need to slightly modify this - precise instructions are in the AI APIs guide (guide 9)

openai = OpenAI()
```

```python
# Create a list of messages in the familiar OpenAI format

messages = [{"role": "user", "content": "Tell me a fun fact"}]
```

```python
# And now - let's ask for a question:

question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
# ask it - this uses GPT 4.1 mini, still cheap but more powerful than nanoresponse = openai.chat.completions.create(    model="gpt-5.4-mini",    messages=messages)question = response.choices[0].message.contentprint(question)
```

```python
# form a new messages list
messages = [{"role": "user", "content": question}]
messages
```

```python
# Ask it again

response = openai.chat.completions.create(
    model="gpt-5.4-mini",
    messages=messages
)

answer = response.choices[0].message.content
print(answer)
```

```python
from IPython.display import Markdown, display

display(Markdown(answer))

```

```python
# Ask the model for a correct answer to compare against

response = openai.chat.completions.create(
    model="gpt-5.4-mini",
    messages=[{"role": "user", "content": f"Answer this question correctly and concisely:\n\n{question}"}]
)

correct_answer = response.choices[0].message.content
print(correct_answer)
```

```python
message = f"""
Here is a question:
{question}

Here is a possible answer that might be correct or incorrect:
{answer}

Please evaluate the answer and provide a score between 0 and 100.

Please provide a correct answer:
{correct_answer}
"""

print(message)
```

```python
messages = [{"role": "user", "content": message}]
messages

response = openai.chat.completions.create(
    model="gpt-5.4",
    messages=messages
)
print(response.choices[0].message.content)

score = response.choices[0].message.content

message = f"""
Here is the score: {score}

Here is the correct answer: {correct_answer}
"""

print(message)
```

```python
# First create the messages:

messages = [{"role": "user", "content": "Pick a business area that might be ripe for an Agentic AI solution."}]

# Then make the first call:

response = openai.chat.completions.create(
    model="gpt-5.4-mini",
    messages=messages
)

# Then read the business idea:

business_idea = response.choices[0].message.content

# And repeat! In the next message, include the business idea within the message

messages = [{"role": "user", "content": f"Pick a pain-point in the following industry: {business_idea}"}]

response = openai.chat.completions.create(
    model="gpt-5.4-mini",
    messages=messages
)

pain_point = response.choices[0].message.content

messages = [{"role": "user", "content": f"Now propose an AI solution to the following pain-point: {pain_point}"}]

response = openai.chat.completions.create(
    model="gpt-5.4-mini",
    messages=messages
)

ai_solution = response.choices[0].message.content

message = f"""
Here is the pain-point: {pain_point}

Here is the AI solution: {ai_solution}
"""

print(message)
```
