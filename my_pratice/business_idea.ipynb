```python
# Import required libraries
from dotenv import load_dotenv
from openai import OpenAI
import os

# Load variables from .env
load_dotenv(override=True)

# Read your Gemini API key
gemini_api_key = os.getenv("GEMINI_API_KEY")

# Create the OpenAI client for Gemini
openai = OpenAI(
    api_key=gemini_api_key,
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/"
)

print("Setup completed successfully!")
```

```python
messages = [
    {
        "role": "user",
        "content": "Choose one business area where Agentic AI could have a significant impact. Respond with only the business area."
    }
]

response = openai.chat.completions.create(
    model="models/gemini-flash-latest",
    messages=messages
)

business_area = response.choices[0].message.content

print(business_area)
```
