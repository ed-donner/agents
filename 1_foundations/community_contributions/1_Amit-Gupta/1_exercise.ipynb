Exercise Lab 1:

First ask the LLM to pick a business area that might be worth exploring for an Agentic AI opportunity. <br>
Then ask the LLM to present a pain-point in that industry - something challenging that might be ripe for an Agentic solution. <br>
Finally have 3 third LLM call propose the Agentic AI solution.

Import dotenv to load environment variables

```python
from dotenv import load_dotenv

load_dotenv(override=True)
```

Load API Key

```python
import os
openai_api_key=os.getenv("OPENAI_API_KEY")

if openai_api_key:
    print(f"OpenAI API Key exists and begins with: {openai_api_key[:8]}")
else:
    print("OPENAI API Key not set - please head to the troubleshooting guide in the setup folder")
```

Import OpenAI

```python
from openai import OpenAI

openai=OpenAI()
```

Get Business area to explore for Agentic AI solution

```python
message="Find a business area that might be worth exploring for an Agentic AI opportunity."
messages=[{"role": "user", "content": message}]
response=openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
industry=response.choices[0].message.content
print(industry)
```

Get Challenge/Pain Point

```python

message=f"""
In the industry of:
{industry}, 

what is some challenging/pain point that might be ripe for an Agentic solution?
"""
messages=[{"role": "user", "content": message}]
response=openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
challenge=response.choices[0].message.content
print(challenge)
```

Get Solution

```python
message=f"""
For industry:
{industry}

and the challenge:
{challenge}

please propose some potential Agentic AI solution.
"""
messages=[{"role": "user", "content": message}]
response=openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
solution=response.choices[0].message.content
print(solution)
```
