```python
from dotenv import load_dotenv
load_dotenv(override=True)

import os
openai_api_key = os.getenv('OPENAI_API_KEY')

from openai import OpenAI
from IPython.display import Markdown, display

openai = OpenAI()

message = [{
    "role": "user",
    "content": """
Respond in the style of Osho.
Imagine Osho meets Buddha.
Ask only one profound question.
Return only the question.
"""
}]

response = openai.chat.completions.create(
    model="gpt-5.4-mini",
    messages=message
)

oshoq = response.choices[0].message.content.strip()
display(Markdown(f"**Osho:** {oshoq}"))


jkreply = f"""
Question from Osho:

{oshoq}

Respond in the style of J. Krishnamurti.
Answer in exactly one sentence.
Return only the answer.
"""

message = [{"role": "user", "content": jkreply}]

response = openai.chat.completions.create(
    model="gpt-5.4-mini",
    messages=message
)

jk = response.choices[0].message.content.strip()
display(Markdown(f"**J. Krishnamurti:** {jk}"))


budhareply = f"""
Question from Osho:

{oshoq}

Answer from J. Krishnamurti:

{jk}

Respond in the style of Buddha.

Rate the answer on:
1. Wisdom
2. Clarity
3. Insight

Return only a single number between 1 and 10.
"""

message = [{"role": "user", "content": budhareply}]

response = openai.chat.completions.create(
    model="gpt-5.4-mini",
    messages=message
)

buddha = response.choices[0].message.content.strip()
display(Markdown(f"**Buddha Rating:** {buddha}/10"))
```
