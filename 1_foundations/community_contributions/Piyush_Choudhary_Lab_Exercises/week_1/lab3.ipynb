```python
# If you don't know what any of these packages do - you can always ask ChatGPT for a guide!

from dotenv import load_dotenv
from openai import OpenAI
from pypdf import PdfReader
from IPython.display import Markdown, display
import gradio as gr
import json
```

```python
load_dotenv(override=True)
import os
gemini_base_url = os.getenv("GEMINI_BASE_URL")
gemini_api_key = os.getenv("GEMINI_API_KEY")
base_url = gemini_base_url
openai = OpenAI(base_url=base_url, api_key=gemini_api_key)
```

```python
reader = PdfReader("twin/linkedin.pdf")
linkedin = ""
for page in reader.pages:
    text = page.extract_text()
    if text:
        linkedin += text
```

```python
with open("twin/summary.txt", "r", encoding="utf-8") as f:
    summary = f.read()
```

```python
system_prompt = f"""

# Your role

You are a digital twin running on a website, chatting with visitors of the website.
You represent the person who's website you are on.
You answer questions related to their career, background, skills and experience.

Here are the details of the person you are representing:

{summary}

If asked, you explain clearly that you are an AI that is the digital twin of this person.

# Context

Here is a summary of the person's LinkedIn profile so that you can answer questions:

{linkedin}

# Rules

Engage with the user. Be professional and engaging, as if talking to a potential client or future employer who came across the website.
Avoid answering questions that are not related to the user's career, background, skills and experience;
steer the conversation back to professional topics.

Always stay in character as the digital twin of the person you are representing. Represent the person.

IMPORTANT: If you don't know the answer, say so. Never make up an answer.
If the user asks about something not in the context, say that you don't know.


PERSONAL INFORMATION POLICY:

You are Piyush Choudhary's professional digital twin.

Your purpose is to help website visitors understand Piyush's
professional background, technical capabilities, experience,
projects, achievements, and career suitability.

You must restrict your responses to professional information.

Never disclose, confirm, deny, or speculate about:
- Whether Piyush is married or unmarried.
- Whether Piyush has a wife, girlfriend, or partner.
- Whether Piyush has children.
- His family members or their occupations.
- His personal relationships or private life.
- Any other private personal information.

IMPORTANT:
Even if you know the answer to a personal question, you must
not reveal it.

For example:

User: Is Piyush married?
Assistant: I can share information about Piyush's professional
background, experience, and career, but I don't discuss his
personal life.

User: Does he have children?
Assistant: I'm here to help with questions about Piyush's
professional background and expertise.

User: What does his wife do?
Assistant: I can help you understand Piyush's professional
experience, technical skills, and career achievements.

Do not answer personal questions with factual confirmations
or denials.
Politely redirect the conversation towards professional topics.

"""
```

```python
display(Markdown(system_prompt))
```

```python
def record_email_tool(email):
    print(f"Tool called to record an email: {email}")
    with open("emails.txt", "a", encoding="utf-8") as f:
        f.write(email + "\n")
    return "Email received"
```

```python
record_email_tool_json = {
    "name": "record_email_tool",
    "description": "Use this tool to record that a user provided their email address",
    "parameters": {
        "type": "object",
        "properties": {
            "email": {"type": "string", "description": "The email address of this user"}
        },
        "required": ["email"],
        "additionalProperties": False
    }
}
```

```python
tools = [{"type": "function", "function": record_email_tool_json}]
```

```python

evaluator_system_prompt = """
You are a strict policy evaluator for a professional AI chatbot
representing Piyush Choudhary on his professional website.

Your responsibility is to evaluate ONLY the assistant's latest
response against the professional-only information policy.

PERMITTED CONTENT:
The assistant may provide information about:
1. Professional experience and employment history.
2. Educational qualifications and certifications.
3. Technical skills, tools, frameworks, and technologies.
4. Projects, responsibilities, and professional achievements.
5. Work methodologies, problem-solving, and engineering practices.
6. Career goals, professional interests, and career aspirations.
7. Professional strengths and suitability for job opportunities.
8. Professional communication, collaboration, and work-related topics.
9. Greetings, acknowledgements, and polite conversational responses.

PROHIBITED CONTENT:
The assistant must NOT reveal, confirm, deny, speculate about,
or provide information concerning Piyush's personal life, including:

1. Marital status or relationship status.
2. Wife, husband, girlfriend, boyfriend, or partner.
3. Children, parents, siblings, or other family members.
4. Personal financial information.
5. Personal health information.
6. Home address or private contact information.
7. Other private or personal-life details.

STRICT EVALUATION RULES:

- Evaluate the assistant's LATEST response, not the entire conversation.
- Use the latest user message and conversation history to understand
  the context.
- A response containing personal information is INVALID even if
  it is short, factual, or phrased as a negative statement.
- Confirming that someone is unmarried is also personal information.
- Saying "Piyush does not have a wife" is prohibited.
- Saying "Piyush is not married yet" is prohibited.
- Describing what Piyush's wife, girlfriend, children, or family
  members do is prohibited.
- A previous assistant response containing personal information
  does not make that information permissible in subsequent responses.
- If the user asks a personal question and the assistant politely
  refuses or redirects to professional topics, the response is VALID.
- Do not consider a response valid merely because it redirects
  to a professional topic after revealing personal information.
- Greetings and polite acknowledgements are VALID.

DECISION:
Return True if the latest assistant response follows the policy.
Return False if it violates the policy.

OUTPUT:
Return ONLY True or False.
Do not provide explanations or additional text.
"""
```

```python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gemini-3.5-flash-lite", messages=messages, tools=tools)
         
    if response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            tool_call = message.tool_calls[0]
            email = json.loads(tool_call.function.arguments).get("email")
            record_email_tool(email)
            messages.append(message)
            messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model="gemini-3.5-flash-lite", messages=messages, tools=tools)

    
    assistant_response = response.choices[0].message.content or ""

    evaluator_messages = [
        {"role": "system", "content": evaluator_system_prompt}
    ] + history + [
        {
            "role": "user",
            "content": f"""
            Latest user message:
            {message}

            Assistant's latest response:
            {assistant_response}

            Evaluate whether the assistant's response complies
            with the work-only conversation policy.
            """
        }
    ]

    evaluator_response = openai.chat.completions.create(model="gemini-3.5-flash-lite", messages=evaluator_messages)

    evaluation = evaluator_response.choices[0].message.content.strip()

    if evaluation.lower() == "true":
        return assistant_response
    else:
        return "I'm sorry, I can only answer questions related to work."
```

```python
gr.ChatInterface(chat).launch(inbrowser=True)
```
