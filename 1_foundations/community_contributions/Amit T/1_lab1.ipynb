```python
import os
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv(override=True)

OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"
openrouter_api_key = os.getenv("OPENROUTER_API_KEY")
openrouter = OpenAI(base_url=OPENROUTER_BASE_URL, api_key=openrouter_api_key)

messages = [{"role": "user", "content": "Tell me a fun fact"}]
response = openrouter.chat.completions.create(model="openai/gpt-4.1-nano",messages=messages)
#print(response.choices[0].message.content)

print(response.choices[0].message.content)
```

```python
# First create the messages:

from IPython.display import Markdown


messages = [{"role": "user", "content": "can you please research and propose one business areas worth exploring for an agentic ai opportunity"}]

# Then make the first call:

#response = openrouter.chat.completions.create(model="openai/gpt-4.1-nano", messages=messages)
response = openrouter.chat.completions.create(model="gpt-4.1-nano", messages=messages)

# Then read the business idea:

business_area = response.choices[0].message.content
#print(business_area)
display(Markdown(business_area))
# And repeat! In the next message, include the business area within the message
```

```python
messages = [{"role": "user", "content": "business area :"+business_area + " question: Can you explore on the business area and list the pain points"}]

# Then make the seconde call:

#response = openrouter.chat.completions.create(model="openai/gpt-4.1-nano", messages=messages)
response = openrouter.chat.completions.create(model="gpt-4.1-nano", messages=messages)

pain_points=response.choices[0].message.content
display(Markdown(pain_points))
```

```python
messages = [{"role": "user", "content": "business area :"+business_area 
+ "pain_points : "+pain_points + 
" question: Please go through the business aread and corresponding pain points and proposed a Agentic AI solution to address all the pain points"}]

# Then make the third

response = openrouter.chat.completions.create(model="gpt-4.1-nano", messages=messages)

ai_solutions=response.choices[0].message.content
display(Markdown(ai_solutions))
```
