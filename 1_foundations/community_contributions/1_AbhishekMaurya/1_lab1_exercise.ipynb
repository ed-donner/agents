Commercial Application:

First ask the LLM to pick a business area that might be worth exploring for an Agentic AI opportunity. <br>
Then ask the LLM to present a pain-point in that industry - something challenging that might be ripe for an Agentic solution. <br>
Finally have 3 third LLM call propose the Agentic AI solution.

```python
from dotenv import load_dotenv

load_dotenv(override=True)
```

```python
import os
google_api_key = os.getenv("GOOGLE_API_KEY")


if google_api_key:
    print(f"GOOGLE API KEY Key exists and begins {google_api_key[:8]}")
else:
    print("GOOGLE API KEY not set - please head to the troubleshooting guide in the setup folder")
```

```python
from openai import OpenAI
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
gemini = OpenAI(base_url=GEMINI_BASE_URL, api_key=google_api_key)
```

```python

message ="pick a business area that might be worth exploring for an Agentic AI opportunity"

messages = [{"role": "user", "content": message}]



response = gemini.chat.completions.create(model="gemini-3.5-flash-lite", messages=messages)
business_area =response.choices[0].message.content
print(business_area)
```

```python
message=f"Business area: {business_area}\n\nIdentify a challenging pain point in this industry that could be suitable for an Agentic AI solution"
messages = [{"role": "user", "content": message}]

response = gemini.chat.completions.create(model="gemini-3.5-flash-lite", messages=messages)
pain_point =response.choices[0].message.content
```

```python
messages = [
    {"role": "user",  "content": f"""Business area: {business_area} Pain point: {pain_point}  Propose an Agentic AI solution that could solve this pain point.
Explain how the agent would work and what value it would provide.
"""
    }
]

response = gemini.chat.completions.create( model="gemini-3.5-flash-lite", messages=messages)

solution = response.choices[0].message.content

print(solution)
```
