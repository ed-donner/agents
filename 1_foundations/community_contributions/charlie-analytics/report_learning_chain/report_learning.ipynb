## Local parsing and remote-call boundary

```python
def replay(stage):
    record = samples[str(stage)]
    print('OFFLINE REPLAY: no new model call')
    print('Exact prompt:', record['prompt'])
    print('Returned model:', record['model'])
    print('Reported cost:', record['cost'])
    print('Finish reason:', record['finish_reason'])
    print('Actual response:', record['reply'])
    return json.loads(record['reply'])

# Optional example only: NOT invoked by any cell.
# Before using it, configure your own key privately, review provider terms,
# verify current model pricing and set a spending limit in your own account.
# A saved example's free price or total spend is not a future cost guarantee.
def live_request(prompt, model):
    import os
    from openai import OpenAI
    with OpenAI(base_url='https://openrouter.ai/api/v1',
                api_key=os.environ['OPENROUTER_API_KEY'],
                max_retries=0, timeout=60) as client:
        # THIS line invokes remote inference. All other example cells are local.
        return client.chat.completions.create(
            model=model, messages=[{'role':'user','content':prompt}],
            max_tokens=1800)
```

## Stage 1: actual observations, checked locally

```python
observations = replay(1)
assert len(observations['observations']) == 3
sources = {r['id']: r for r in reports}
assert {o['report_id'] for o in observations['observations']} == set(sources)
for observation in observations['observations']:
    card = sources[observation['report_id']]
    assert observation['basis'] == 'researcher-supplied source card'
    assert observation['evidence_type'] == 'researcher_note'
    assert any(note['page'] == observation['source_page'] and
               note['note'] == observation['evidence_note']
               for note in card['researcher_notes'])
print('Three supplied-note/page references match. Interpretation still needs review.')
```

## Stage 2: actual principles, checked locally

```python
principles = replay(2)
observation_ids = {o['id'] for o in observations['observations']}
assert len(principles['principles']) == 3
for principle in principles['principles']:
    assert principle['observation_ids']
    assert set(principle['observation_ids']) <= observation_ids
assert principles['truthfulness_applies_to_all_genres'] is True
print('Three principles reference known observations; truthfulness applies to all genres.')
```

## Stage 3: actual paid final-stage comparison
The final prompt uses distilled structural principles, without raw cards, to reduce confusion between learning examples and evidence. It proposes a 1,200-word guide; it does not generate a complete 1,200-word report.

```python
draft = replay(3)
principle_ids = {p['id'] for p in principles['principles']}
assert len(draft['sections']) == 5
assert [s['word_allocation'] for s in draft['sections']] == [180,260,280,260,220]
assert sum(s['word_allocation'] for s in draft['sections']) == 1200
for section in draft['sections']:
    assert section['evidence_needed'] and section['principle_rationale']
    assert set(section['principle_ids']) <= principle_ids
assert draft['factual_claim_ids'] == []
raw_word_count = len(draft['opening'].split())
print('Raw opening words:', raw_word_count)
print('Requested 80–100 words:', 80 <= raw_word_count <= 100)
assert raw_word_count == 75  # Honest regression check for this saved sample.
```

## Local editorial repair, not a model response
The paid model's raw opening has 75 words and fails the requested 80–100. The local edit adds one sentence to reach 85. The original response remains unchanged in the fixture. Section 1's 180-word allocation includes the opening.


```python
added_sentence = ' Editors should document unresolved questions before deciding whether to publish.'
edited_opening = draft['opening'] + added_sentence
assert len(edited_opening.split()) == 85
print('LOCALLY EDITED opening, 85 words:', edited_opening)
print('Added locally:', added_sentence.strip())
```
