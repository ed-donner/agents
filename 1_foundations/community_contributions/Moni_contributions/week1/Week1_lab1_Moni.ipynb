```python
import os
import getpass
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

#---------------------------------------
# API Keys
#---------------------------------------

os.environ["OPENROUTER_API_KEY"] = getpass.getpass(
    "Enter OpenRouter API Key: "
)

OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")


#---------------------------------------
# Initialise LLM
#---------------------------------------

llm = ChatOpenAI(
    model="openai/gpt-4o-mini",
    temperature=0.3,
    max_tokens=500,
    openai_api_key=OPENROUTER_API_KEY,
    base_url="https://openrouter.ai/api/v1"
)


#---------------------------------------
# Step 1
# Pick an Industry
#---------------------------------------

def pick_business_area():

    prompt = """
    You are an expert in AI strategy.

    Suggest ONE business domain that has significant
    opportunities for Agentic AI adoption.

    Requirements:
    - Must have complex workflows.
    - Must involve multiple stakeholders.
    - Must have repetitive or decision-making tasks.
    - Should be realistic for an Agentic AI solution.

    Return:
    - Industry Name
    - Why it is suitable for Agentic AI
    """

    response = llm.invoke([HumanMessage(content=prompt)])

    return response.content


#---------------------------------------
# Step 2
# Find pain points
#---------------------------------------

def find_pain_points(industry):

    prompt = f"""
    The following industry has been selected:

    {industry}

    Identify the top pain points that are difficult
    to solve using traditional automation.

    Focus on:
    - inefficiencies
    - manual decision making
    - fragmented systems
    - bottlenecks

    Explain:
    - the problem
    - who is affected
    - why current solutions are inadequate
    """

    response = llm.invoke([HumanMessage(content=prompt)])

    return response.content


#---------------------------------------
# Step 3
# Propose Agentic AI solution
#---------------------------------------

def propose_agentic_solution(industry, pain_points):

    prompt = f"""
    Industry:
    {industry}


    Pain Points:
    {pain_points}


    Design an Agentic AI solution.

    Describe:

    1. What agents are required.
    2. What each agent does.
    3. How agents collaborate.
    4. Which tools they might use.
    5. Expected business benefits.
    6. Why Agentic AI is preferable over a simple chatbot.

    Make the solution practical and realistic.
    """

    response = llm.invoke([HumanMessage(content=prompt)])

    return response.content


#---------------------------------------
# Run Pipeline
#---------------------------------------

industry = pick_business_area()

print("\n=========== INDUSTRY ===========\n")
print(industry)


pain_points = find_pain_points(industry)

print("\n=========== PAIN POINTS ===========\n")
print(pain_points)


solution = propose_agentic_solution(
    industry,
    pain_points
)

print("\n=========== AGENTIC AI SOLUTION ===========\n")
print(solution)
```
