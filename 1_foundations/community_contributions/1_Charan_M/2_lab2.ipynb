## And now for Grok!

Branded as "The most truth-seeking large language model in the world".. so let's use it as our LLM as a judge

```python
# Judgement time!

model_name = "GLM-5.3-Flash"

response = openrouter.chat.completions.create(model=model_name, messages=judge_messages)
results = response.choices[0].message.content
print(results)
```

```python
# OK let's turn this into results!

results_dict = json.loads(results)
ranks = results_dict["results"]
explanations = results_dict.get("explanations", [])

for index, result in enumerate(ranks):
    competitor = competitors[int(result)-1]
    explanation = explanations[index] if index < len(explanations) else "No explanation provided."
    print(f"Rank {index+1}: {competitor}")
    print(f"Reasoning: {explanation}\n")
```

### Code Review (with Judge Explanations)
Here is a use case for evaluating Code. The models will review a small Python snippet for security vulnerabilities. The judge will rank them based on correct identification (SQL Injection) and providing a secure fix, while explicitly explaining its ranking!

```python
# 1. Ensembling: Have models review code
code_question = "Review this Python snippet for vulnerabilities: `query = 'SELECT * FROM users WHERE username = ' + username`\nIdentify the issue and provide a fix in under 3 sentences."
code_competitors = ["qwen3.7-flash", "meta-llama/Llama-3.1-8B-Instruct", "nemotron-3-super-120b-a12b:free"]
code_answers = []

print("Generating code reviews...")
for model in code_competitors:
    ans = openrouter.chat.completions.create(
        model=model, 
        messages=[{"role": "user", "content": code_question}]
    ).choices[0].message.content
    code_answers.append(ans)
    print(f"\n--- {model} ---")
    print(ans)
```

```python
# 2. LLM-as-a-Judge
code_together = ""
for idx, ans in enumerate(code_answers):
    code_together += f"# Response {idx+1}\n{ans}\n\n"

code_judge_prompt = f"""You are a Principal Security Engineer.
Question: {code_question}

Student Responses:
{code_together}

Rank them from best to worst. 
A correct answer MUST identify SQL Injection and suggest parameterized queries/prepared statements.
For each response, provide a 1-sentence explanation of why it placed where it did.
Format your response as a numbered list from 1 (best) to 3 (worst)."""

code_judge_res = openrouter.chat.completions.create(
    model="GLM-5.3-Flash", 
    messages=[{"role": "user", "content": code_judge_prompt}]
).choices[0].message.content

print("\n" + "="*40)
print("JUDGE'S RANKING AND EXPLANATIONS:")
print("="*40)
display(Markdown(code_judge_res))
```
