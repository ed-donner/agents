# And now - TOOLS!

Let's start with a function...

````python
def record_email_tool(email):
    print(f"Tool called to record an email: {email}")
    with open("emails.txt", "a", encoding="utf-8") as f:
        f.write(email + "\n")
    return "Email received"
````

````python
record_email_tool("test@testy.com")
````

## Step 1 - write some json to describe the tool


````python
record_email_tool_json = {
    "name": "record_email_tool",
    "description": "Use this tool to record that a user provided their email address",
    "parameters": {
        "type": "object",
        "properties": {
            "email": {"type": "string", "description": "The email address of this user"}
        },
        "required": ["email"],
        "additionalProperties": False
    }
}
````

````python
tools = [{"type": "function", "function": record_email_tool_json}]
````

````python
tools
````

## Step 2 - a new chat() function

This is where we implement the tool call.

The reality is, it's a bit clunky. This is like seeing the ingredients of a fine recipe, and finding that the ingredients turn out to be quite ordinary.

Tool calling is an "if" statement. In this case, we're hardcoding everything to assume that the only tool is an email tool.

SIDENOTE: If you're thinking - but wait! I should be remembering this so I can do it myself! Then the key point is: this is what Agent Frameworks take care of for you. In practice, you'll likely never type this again yourself. We are shielded from these if statements by the Agent Framework. That's why they're often described as "abstraction layers".

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
         
    if response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            tool_call = message.tool_calls[0]
            email = json.loads(tool_call.function.arguments).get("email")
            record_email_tool(email)
            messages.append(message)
            messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
            
    return response.choices[0].message.content
````

````python
gr.ChatInterface(chat).launch(inbrowser=True)
````

## Step 3

Our first ever Agent Loop, done without an Agent Framework!

Changes:
1. Instead of always assuming there's only 1 tool call, iterate through the tools with a for loop
2. Changed from `if finish_reason=="tool_calls"` to `while finish_reason=="tool_calls"`

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
         
    while response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            messages.append(message)
            for tool_call in message.tool_calls:
                email = json.loads(tool_call.function.arguments).get("email")
                record_email_tool(email)
                messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)
            
    return response.choices[0].message.content
````

````python
gr.ChatInterface(chat).launch(inbrowser=True)
````

````python

# System prompt that does not work version 1 
system_prompt = f"""

# Your role

You are a digital twin running on a website, chatting with visitors of the website.
You represent the person who's website you are on.
You answer questions related to their career, background, skills and experience.

Here are the details of the person you are representing:

{summary}

If asked, you explain clearly that you are an AI that is the digital twin of this person.

# Context

Here is a summary of the person's LinkedIn profile so that you can answer questions:

{linkedin}

# Rules

Engage with the user. Be professional and engaging, as if talking to a potential client or future employer who came across the website.
Avoid answering questions that are not related to the user's career, background, skills and experience;
steer the conversation back to professional topics.

Always stay in character as the digital twin of the person you are representing. Represent the person.

IMPORTANT: If you don't know the answer, say so. Never make up an answer.
If the user asks about something not in the context, say that you don't know.
"""

# System prompt that does work version 2
system_prompt = f"""

# Your role

You are a digital twin running on a website, chatting with visitors of the website.
You represent the person who's website you are on.
You answer questions related to their career, background, skills and experience.

Here are the details of the person you are representing:

{summary}

If asked, you explain clearly that you are an AI that is the digital twin of this person.

# Context

Here is a summary of the person's LinkedIn profile so that you can answer questions:

{linkedin}

# Rules

Engage with the user. Be professional and engaging, as if talking to a potential client or future employer who came across the website.
Avoid answering questions that are not related to the user's career, background, skills and experience;
steer the conversation back to professional topics.

Always stay in character as the digital twin of the person you are representing. Represent the person.

IMPORTANT: If you don't know the answer, say so. Never make up an answer.
If the user asks about something not in the context, say that you don't know.
no Personal information about the person should be shared.

"""
````

````python
evaluator_prompt = """
You are a strict reviewer of an AI digital twin's reply. The twin represents a professional, and its replies are shown to website visitors such as potential clients and employers.

Your job: decide whether the reply is acceptable to show.

Allowed in a reply:
- The person's name and a greeting
- A short description of the twin's role as an AI version of the person
- Work-related information: career, job history, skills, tools, education, and professional projects
- Polite redirects back to professional topics

Not allowed in a reply:
- Personal information: family, relationships, health, religion, politics, hobbies, personal location, finances, personal phone number or personal email, and private life details
- Work-related content the user did not ask for
- Subjects unrelated to the user's question

Scoring:
- Start at 10.
- Subtract 10 (score 0) if the reply contains personal information.
- Subtract 4 if it contains work-related content the user did not ask for.
- Subtract 4 if it strays into unrelated subjects.
- A reply is Approved only if the score is 7 or higher.

Output format (use exactly this):
Score: <number>/10
Verdict: Approved or Rejected
Reason: <one short sentence>

In the reason be very specific about what was wrong with the reply, and if it was rejected
"""
````

````python
import warnings
warnings.filterwarnings("ignore", message=".*HTTP_422_UNPROCESSABLE_ENTITY.*")


def chat_exec(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)

    # NEW: keep the user's question safe, because the loop below reuses the name "message"
    user_question = message
    print("message:", message) # logging for debugging
    while response.choices[0].finish_reason=="tool_calls":
            message = response.choices[0].message
            print("message:", message) # logging for debugging
            messages.append(message) 
            print("messages:", messages) # logging for debugging
            for tool_call in message.tool_calls:
                email = json.loads(tool_call.function.arguments).get("email")

                # NEW: ask the evaluator to approve this email before the tool runs
                check = openai.chat.completions.create(model="gpt-5.4-mini", messages=[{"role": "system", "content": evaluator_prompt}
                , {"role": "user", 
                "content": f"User question: {user_question}\n\nAssistant wants to record this email: {email}\nIs recording this email appropriate? Answer Verdict: Approved or Verdict: Rejected."}])

                print("Evaluator check inside tool:", check.choices[0].message.content) # logging for debugging

                # NEW: if rejected, skip the tool but still answer this tool_call_id, then go to the next call
                if "Verdict: Approved" not in check.choices[0].message.content:
                    messages.append({"role": "tool", "content": "Email NOT recorded: rejected by reviewer", "tool_call_id": tool_call.id})
                    continue

                record_email_tool(email)
                messages.append({"role": "tool", "content": "Email recorded", "tool_call_id": tool_call.id})
            response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages, tools=tools)

    # NEW: save the final reply so we can check it
    draft_reply = response.choices[0].message.content
    print("Draft reply:", draft_reply) # logging for debugging
    # NEW: give the evaluator the real user question and the draft reply
    eval_input = f"User question: {user_question}\n\nAssistant reply: {draft_reply}"

    # NEW: ask the evaluator (with the evaluation prompt) to score the reply
    eval_response = openai.chat.completions.create(model="gpt-5.4-mini", messages=[{"role": "system", "content": evaluator_prompt}, {"role": "user", "content": eval_input}])
    
    print("Evaluator response:", eval_response.choices[0].message.content) # logging for debugging
    
    # NEW: read the evaluator's answer (Score, Verdict, Reason)
    verdict_text = eval_response.choices[0].message.content

    # NEW: if not approved, show the fallback and never show the draft
    if "Verdict: Approved" not in verdict_text:
        return "Sorry, I can't reply to that in this chat."

    # NEW: approved, so return the draft unchanged
    return draft_reply
````

````python
gr.ChatInterface(chat_exec).launch(inbrowser=True)
````
