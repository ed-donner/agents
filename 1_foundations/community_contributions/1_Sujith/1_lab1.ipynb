```python
# Check the key - if you're not using OpenAI, check whichever key you're using! Ollama doesn't need a key.

import os
from openai import OpenAI
from IPython.display import Markdown, display
```

```python

openai_api_key = os.getenv('OPENAI_API_KEY')
openai = OpenAI()
```

```python

messages = [{"role": "user", "content": "Pick a business area and propose a business idea in that area for agentic AI"}]
print("Called the API to get a business area for using Agentic AI")
response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
business_area = response.choices[0].message.content
print("We got business area, now let's get a painpoint in that area\n")

messages = [{"role": "user", "content": f"Based on the business area '{business_area}', Present a painpoint in the industry that might be solved by agentic AI."}]
print("Called the API to get a painpoint in the business area")
response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
business_painpoint = response.choices[0].message.content
print("We got the painpoint\n")

messages = [{"role": "user", "content": f"Based on the business painpoint '{business_painpoint}', Present a agentic AI solution to the painpoint."}]
print("Called the API to get a solution to the painpoint")
response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
business_solution = response.choices[0].message.content
print("We got the solution to the painpoint and \n\nplease run the next cell to display the responce in a more readable format")
```

```python
messages = [{"role": "user", "content": f"Summarize the below topics in such a way that propose a business plan and for that business plan, summarize the painpoint mentioned then provide a solution to the painpoint using below context. Dont introduce new stuff just do it based on given below context and make it compact, ignore unnecessary information and include only the most relevant details. \n\nBusiness Area: {business_area}\n\nPainpoint: {business_painpoint}\n\nSolution: {business_solution}"}]
response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
final = response.choices[0].message.content
display(Markdown(final))
```
