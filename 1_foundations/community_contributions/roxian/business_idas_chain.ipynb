# exersise

first ask the LLM to pick a business area that might be worth exploring for an agentic ai oppertunity

then ask the LLM to present a paint-point in that industry, something challenging that mightb be ripe for an agentic solution

finnaly have third LLM call propose the agentic ai sollution

```python
# Check the key - if you're not using OpenAI, check whichever key you're using! Ollama doesn't need a key.

import os
from openai import OpenAI

# Initialize the OpenAI client pointing to Groq's server
client = OpenAI(
    api_key=os.environ.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1",
)
```

iteration 1 

```python
question="pick up a business area that might be worth exploring for an agentic oppertunity and can be fixed with ai or can be simplify with ai agent"

message1=[{"role":"user","content":question}]
```

```python
response=client.chat.completions.create(model="llama-3.3-70b-versatile",messages=message1)
ideas=response.choices[0].message.content
print(ideas)
```

iteration 2

```python
messages2=[{"role":"user","content":f"this is the {ideas} some ideas your job is to check how can it be solved with agentic ai present a paint-point in that industry, something challenging that mightb be ripe for an agentic solution here are the ideas {ideas}"}]

messages2
```

```python
response2=client.chat.completions.create(model="llama-3.3-70b-versatile",messages=messages2)
illustration=response2.choices[0].message.content
print(illustration)
```

iteration 3

```python
messages3=[{"role":"user","content":f"now acording to these illustrations {illustration} propose the business ideas in detail"}]

messages3
```

```python
response3=client.chat.completions.create(model="openai/gpt-oss-20b",messages=messages3)
finnalproposal=response3.choices[0].message.content
print(finnalproposal)
```
