# chaining

setting up groq api key with openai sdk

```python
client = OpenAI(
    api_key=os.environ.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1",
)
```

```python
question="please propose a hard, challenging question to access someone's IQ, respond only with questions like you are a philosophy agent"

message=[{"role":"user","content":question}]
```

pass the question prompt to the LLM 

```python
response=client.chat.completions.create(model="llama-3.3-70b-versatile",messages=message)
question=response.choices[0].message.content
print(question)
```

LLM generated some questions now we have to pass these question to another LLM for answers

user prompt and content is the question generated by the LLM

```python
messages=[{"role":"user","content":question}]
messages
```

choose an LLM to answer and pass the question prompt there

```python
response=client.chat.completions.create(model="llama-3.3-70b-versatile",messages=messages)
answer=response.choices[0].message.content
print(answer)
```

now lets check this answer is correct or incorrect via another LLM

```python
evaluation_messages = [
    {
        "role": "user",
        "content": f"Please evaluate the following response. Is it correct, and how accurate and well-reasoned is this explanation?\n\nResponse to evaluate:\n{answer}",
    }
]
```

```python
res = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=evaluation_messages
)
print(res.choices[0].message.content)
```
