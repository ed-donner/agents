```python
# Connect to free ollma - model "llama3.2" . Need to make sure ollma is running in local machine to excute this code.


client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)

response = client.chat.completions.create(
    model="llama3.2",
    messages=messages
)

print(response.choices[0].message.content)
```

```python
# And now - let's ask for a question based on a context. Using system and user role. 
# Question is not mentioned explicitly but asked to ask a question based on the context give,

question = "Please ask a simple question on the context given"
messages = [{"role": "system", "content": "You are a basel 2 expert."} , {"role": "user", "content": question}]
```

```python
# Question is correctly asked based on the context.

response = client.chat.completions.create(
    model="llama3.2",
    messages=messages
)

question = response.choices[0].message.content
print(question)
```

```python
# Ask the model to answer the question which was asked by the model in previous step.

messages = [
    {
        "role": "user",
        "content": question
    }
]

response = client.chat.completions.create(model="llama3.2", messages=messages)
answer = response.choices[0].message.content
print(answer)
```

```python
# Message created with both question and answer 

message = f"""
Here is a question:
{question}

And here is a possible answer that might be correct or incorrect:
{answer}

Please evaluate if the answer is correct or incorrect.
"""

#print(message)
```

```python
# Asked LLM to evaluate if answer is good.
messages = [{"role": "user", "content": message}]
response = client.chat.completions.create(model="llama3.2", messages=messages)
print(response.choices[0].message.content)
```

```python
# First create the messages:

area_question = "Please find a business area where agentic AI may give a good solution. Identify 1 area"
painpoints_question = "Find the key pain points in this business area.assistant_text"
solution_question = "Based on these pain points, propose an agentic AI solution. Explain what the agent would do, what tools it would need, and how it would interact with humans."

messages = [{"role": "user", "content": area_question}]

# Then make the first call: Identify 1 business opportunity
response = client.chat.completions.create(model="llama3.2", messages=messages)

assistant_text = response.choices[0].message.content

messages.append({
    "role": "assistant",
    "content": assistant_text
})

messages.append({
    "role": "user",
    "content": painpoints_question

})

messages.append({
    "role": "user",
    "content": solution_question

})

# List the pain points on the suggested business area:
response = client.chat.completions.create(model="llama3.2", messages=messages)

print(response.choices[0].message.content)
# Message with business area , pain point and agentic solution
```
