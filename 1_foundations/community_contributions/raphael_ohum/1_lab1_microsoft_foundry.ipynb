```python
# Check the key - if you're not using OpenAI, check whichever key you're using! Ollama doesn't need a key.

import os
azure_api_key = os.getenv('AZURE_API_KEY')

if azure_api_key:
    print(f"Azure API Key exists and begins {azure_api_key[:8]}")
else:
    print("OpenAI API Key not set - please head to the troubleshooting guide in the setup folder")
    
```

```python
# And now - the all important import statement
# If you get an import error - head over to Q5 and Q25 in the FAQ at https://edwarddonner.com/avatar
# Even for other LLM providers like Gemini, you still use this OpenAI import - see Guide 9 for why

from openai import OpenAI
```

```python
# And now we'll create an instance of the OpenAI class
# If you're not sure what it means to create an instance of a class - head over to the guides folder (guide 6)!
# If you get a NameError - head over to the guides folder (guide 6)to learn about NameErrors - always instantly fixable
# If you're not using OpenAI, you just need to slightly modify this - precise instructions are in the AI APIs guide (guide 9)

# You need to have created a Microsoft Foundry resource in Azure, a Project under the resource, and a deployment of the model you want to use. 
endpoint = "https://agenticailearn.services.ai.azure.com/openai/v1"

# And now we'll define the deployment names for the models behind the endpoint
deployment_name_gpt_5_mini = "gpt-5-mini"
deployment_name_gpt_4_1_nano = "gpt-4.1-nano"

openai = OpenAI(api_key=azure_api_key, base_url=endpoint)
```

```python
# Create a list of messages in the familiar OpenAI format

messages = [{"role": "user", "content": "Tell me a fun fact"}]
```

```python
messages
```

```python
# And now call it! Any problems, head to the troubleshooting guide
# This uses GPT 5.4 nano, an incredibly cheap model
# The APIs guide (guide 9) has exact instructions for using even cheaper or free alternatives to OpenAI
# If you get a NameError, head to the guides folder (guide 6) to learn about NameErrors - always instantly fixable

response = openai.chat.completions.create(model=deployment_name_gpt_4_1_nano, messages=messages)
print(response.choices[0].message.content)
```

```python
# And now - let's ask for a question:

question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
# ask it - this uses GPT 5.4 mini, still cheap but more powerful than nano

response = openai.chat.completions.create(model=deployment_name_gpt_4_1_nano, messages=messages)
question = response.choices[0].message.content
print(question)
```

```python
# form a new messages list

messages = [{"role": "user", "content": question}]
messages
```

```python
# Ask the model to answer the hard question

response = openai.chat.completions.create(model=deployment_name_gpt_4_1_nano, messages=messages)
answer = response.choices[0].message.content
print(answer)
```

```python
from IPython.display import Markdown, display

display(Markdown(answer))
```

```python
message = f"""
Here is a question:
{question}

And here is a possible answer that might be correct or incorrect:
{answer}

Please evaluate if the answer is correct or incorrect.
"""

print(message)
```

```python
messages = [{"role": "user", "content": message}]
response = openai.chat.completions.create(model=deployment_name_gpt_5_mini, messages=messages)
print(response.choices[0].message.content)
```

```python
# First create the messages:
user_message = "You are an expert business idea generator and seasoned entrepreneur. Please propose an industry that is worth exploring for an impactful Agentic AI business/startup opportunity."
messages = [{"role": "user", "content": user_message}]

# Then make the first call:

response = openai.chat.completions.create(model=deployment_name_gpt_5_mini, messages=messages)

# Then read the business idea:

business_area = response.choices[0].message.content

print(business_area)

# And repeat! In the next message, include the business area within the message
```

```python
# First create the messages:
user_message = f"""
Below is the proposed business area for an impactful Agentic AI business/startup opportunity:
{business_area}

Review the pain points and establish if these are valid pain points that are valuable and challenging enough to solve. 
If they are, determine if they are ripe for an Agentic AI solution. 
If they are not, propose a new pain-point in this industry that is valuable and challenging enough to solve, as well as ripe for an Agentic AI solution.
"""
messages = [{"role": "user", "content": user_message}]

# Then make the first call:

response = openai.chat.completions.create(model=deployment_name_gpt_5_mini, messages=messages)

# Then read the pain points:

pain_points = response.choices[0].message.content

print(pain_points)

# And repeat! In the next message, include the business area within the message
```

```python
# First create the messages:
user_message = f"""
Clarify the solution proposed in {business_area}. 

Make sure that the solution is feasible, scalable, has a clear path to market, and sufficiently addresses the identified pain points in {pain_points}.

Keep your response concise, clear, and simple using principles like KISS and ELI5.
"""
messages = [{"role": "user", "content": user_message}]

# Then make the first call:
response = openai.chat.completions.create(model=deployment_name_gpt_5_mini, messages=messages)

# Then read the solution:
solution = response.choices[0].message.content

print(solution)
```

```python
_business_area = f"### Business Area\n{business_area}"
_pain_points = f"### Pain Points\n{pain_points}"
_solution = f"### Solution\n{solution}"

display(Markdown(_business_area))

display(Markdown(_pain_points))

display(Markdown(_solution))
```

### Business Area
Industry to explore: Construction project lifecycle (commercial, infrastructure, and heavy civil) — Agentic AI for end-to-end field orchestration and automation

Why this industry
- Massive market and pain: Global construction is ~10T+/yr, highly fragmented, chronically late/over budget (average 20–45% cost overruns on many project types). Labor shortages, material waste, safety incidents and poor documentation are persistent problems.
- Digitization gap: Many firms use BIM, Procore, Primavera, etc., but workflows remain manual, reactive and distributed across subs, foremen, suppliers and inspectors.
- High leverage for automation: Much of the work is coordination, repeated decision-making, compliance checks and visual inspection — tasks well suited to autonomous agents that can observe, decide and act across physical and digital systems.

How agentic AI creates value (examples)
- Autonomous schedule and subcontractor coordination agent: reads project schedule/BIM, senses delays, negotiates revised timings with subs via messaging/phone, re-sequences downstream work and updates stakeholders.
- Visual inspection agent: ingests drone/phone imagery + plans, detects defects (wrong rebar, misalignments), files annotated RFIs and routes for approval.
- Procurement and vendor negotiation agent: enumerates material needs from BIM, sources quotes, negotiates price/delivery, issues POs and tracks deliveries.
- Safety and compliance agent: monitors sensor/camera feeds and daily reports, flags unsafe conditions, generates compliance logs and arranges remedial actions.
- Digital twin controller: maintains a live state of the build (BIM + schedules + material + labor), runs simulations and recommends optimal resource allocation to hit milestones with least cost.

Business model
- SaaS platform (per-project or per-seat) for core orchestration and agents.
- Transaction/marketplace fees for procurement and equipment rental automation.
- Premium enterprise modules (BIM integrations, custom agents, managed deployment).
- Pilot/POC pricing to prove ROI (success-based fees tied to schedule/cost improvements).

MVP recommendation (focus and scope)
- Target customer: mid-size general contractors (GCs) and specialty contractors on projects $5M–$100M where digitization exists but coordination is still manual.
- Start with 1–2 use-cases to get traction:
  1) Schedule/reschedule + subcontractor negotiation agent that integrates with Procore/Primavera and communicates via SMS/WhatsApp/Teams.
  2) Visual inspection agent for early-structure phases (rebar/formwork) using phone imagery and simple drones.
- Deliverables: agent dashboard, integration connectors, mobile UI for field approvals, audit log for decisions.

Differentiation
- True agentic behavior: agents not only recommend but execute (negotiate, schedule, issue POs) within guardrails and auditable logs.
- Closed loop with physical assets: drone/CV + BIM + real-time schedule orchestration versus point-solution inspection or analytics-only players.
- Domain knowledge and workflows: pre-built construction ontologies, safety rules, contract/permit templates.

Technical considerations
- Multi-agent orchestration architecture with persistent project state (knowledge graph/digital twin).
- LLMs fine-tuned on construction docs, contracts, change orders; CV models for defect detection in outdoor, occluded contexts.
- Integrations: Procore, Autodesk/BIM 360, Primavera/MS Project, accounting systems, supplier APIs, SMS/WhatsApp/Teams.
- Safety/guardrails: human-in-the-loop approvals for dangerous or high-value actions; role-based permissions and immutable audit logs.
- Data strategy: connectors for plans/BIM, drone/photo ingestion, historical project data; synthetic data augmentation for CV and negotiation scenarios.

Risks and mitigations
- Liability and safety: never let an agent perform unsupervised high-risk physical actions; always require sign-off for safety-critical decisions; carry professional indemnity and partner with insurers.
- Adoption/trust: start as “assistant+executer with approval” and show clear time/money savings; build transparency (why an agent recommended X).
- Data privacy & IP: secure on-prem or VPC deployments, strict access controls; offer data residency for enterprise customers.
- Regulatory/permits: integrate permit workflows and local code checkers, but keep legal review in the loop.

Go-to-market / distribution
- Pilot partnerships: 2–3 mid-size GCs on 6–12 month pilots tied to concrete ROI metrics (days saved, cost reductions).
- Channel partners: integration/partnership with Procore, Autodesk, Trimble; drone inspection firms and equipment rental companies.
- Sales motion: solutions selling via construction-focused reps; proof-of-value pilots funded partially by vendor to prove savings.
- Marketing: case studies showing reduced RFIs, avoided delays, shortened procurement cycles.

Metrics that matter
- Reduction in schedule slippage % (weeks saved per project)
- Reduction in RFIs and average time to resolve
- Material cost savings / fewer change orders
- Time saved per PM/foreman per week
- ARR, pilot-to-paid conversion, customer retention

Team & hiring (first 6–12 months)
- Construction domain lead (ex-GC PM or PM with field credibility)
- ML engineers (CV + LLM fine-tuning) and a backend systems engineer
- Integrations engineer (BIM/Procore/ERP connectors)
- Product manager with construction workflow experience
- Sales lead with construction tech relationships

Early experiments to validate quickly (0–3 months)
1) Scheduling negotiation pilot: build a lightweight agent that proposes 1–3 rescheduling options and sends messages to subs; measure acceptance rate, time saved and downstream impact.
2) Photo-based defect detection prototype: collect phone/drone images from 3 jobs, run off-the-shelf CV models fine-tuned to detect 2–3 common defects (rebar spacing, missing formwork) and generate annotated RFIs. Compare to human inspectors.
3) Procurement quick-win: automate sourcing/quoting for 5 high-volume materials on one pilot project, measure cost and lead-time improvements.
4) Daily report automation: auto-generate daily logs and progress photos summary from field inputs and measure PM time saved.
5) Safety alert pilot: simple rules-based agent that watches checklists + timecard data and surfaces likely non-compliance events to safety managers.

Why now
- LLM and CV capabilities + agent orchestration frameworks make autonomous coordination feasible.
- Pressure on margins and labor shortages accelerate adoption of automation.
- Construction companies are increasingly buying SaaS and connecting digital tools — integration points exist to embed agents.

Potential exit / strategic acquirers
- Autodesk, Procore, Trimble, large EPC firms, major insurers or equipment rental companies.

Bottom line
Construction is a large, inefficient market with repeated, costly coordination problems that map directly to agentic AI strengths (observe, decide, act across many stakeholders). A focused product that automates scheduling/coordination + visual inspection + procurement, delivered as auditable, human-supervised agents, can rapidly prove ROI, scale across projects, and unlock major labor, cost and safety improvements.
