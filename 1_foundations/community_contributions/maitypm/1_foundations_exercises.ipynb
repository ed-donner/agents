```python
# Load the .env file

from dotenv import load_dotenv
load_dotenv(override=True)
```

```python
# Check if the OpenAI API key is set

import os 
openai_api_key = os.getenv("OPENAI_API_KEY")

if openai_api_key:
    print(f"OpenAI API key exists and begins with {openai_api_key[:8]}")
else:
    print("OpenAI API key not set.")
```

```python
# Initiate OpenAI

from openai import OpenAI 
openai = OpenAI()
```

# TWIN - LAB 1

```python
# Create the first message - ask to pickup a business area
biz_area = """Pickup a business area in a digital company where you can apply AI.
Give only the business area name along with the company type, no other text or explanation.
"""
messages = [{"role": "user", "content": biz_area}]

# Make the first model call
biz_response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
biz_area = biz_response.choices[0].message.content
print(biz_area)
```

```python
# Create the second message - ask to pickup a pain area in this particular business area

pain_area = f"""Pickup a pain area in the business area of {biz_area}.
Give pain area name in not more than 10 words, and brief explanation about it in not more than 50 words.
Include the business area and company type in the response.
"""
messages = [{"role": "user", "content": pain_area}]

# Make the second model call
pain_response = openai.chat.completions.create(model="gpt-5.4-mini", messages=messages)
pain_area = pain_response.choices[0].message.content 
print(pain_area)
```

```python
# Create the third message - ask solution around the pain area from pov of Agentic AI

from IPython.display import Markdown, display

agentic_ai = f"""
Here is a business area and a pain point that I would like you to look at:
{pain_area}

Give me solution around this pain area that uses Agentic AI. Structure the solution such that it can be directly presented to senior management.
Your response should be brief but on point. Response should be not more than 300 words.
"""
messages = [{"role": "user", "content": agentic_ai}]

# Make the final model call
aa_resp = openai.chat.completions.create(model="gpt-6-luna", messages=messages)
agentic_ai = aa_resp.choices[0].message.content

display(Markdown(agentic_ai))
```
