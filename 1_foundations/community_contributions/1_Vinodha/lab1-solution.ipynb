```python
from dotenv import load_dotenv
load_dotenv(override=True)

import os
openai_api_key = os.getenv('OPENAI_API_KEY')

from openai import OpenAI

openai = OpenAI()

messages = [{"role": "user", "content": "Pick a business area that might be exploring for an agentic AI opportunity. Give answer in a single line"}]

response = openai.chat.completions.create(model="gpt-5.4", messages=messages)
business_idea = response.choices[0].message.content
print("BUSINESS_IDEA:", business_idea)

message = f"""
            Here is the business area {business_idea}
            Present a pain point in this business area - something challenging that might ripe for an agentic solution.
            Give answer in few sentences
            """

messages = [{"role": "user", "content": message}]            
          
response = openai.chat.completions.create(model="gpt-5.4", messages=messages)
pain_point = response.choices[0].message.content
print("PAIN_POINT:", pain_point)

message = f"""
            Here us the business area {business_idea}
            And, here is the pain point for this business {pain_point}
            Present a AI solution for this pain point in this business area
            """

messages = [{"role": "user", "content": message}]
            
            
response = openai.chat.completions.create(model="gpt-5.4", messages=messages)
ai_solution = response.choices[0].message.content
print(ai_solution)
```
