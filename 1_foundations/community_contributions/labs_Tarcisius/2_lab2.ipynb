## And now for Grok!

Branded as "The most truth-seeking large language model in the world".. so let's use it as our LLM as a judge

```python
# Judgement time!
# Grok is "The most truth-seeking large language model in the world."

judge_model_name = "gpt-5.4-nano"

response = openai.chat.completions.create(model=judge_model_name, messages=judge_messages)
results = response.choices[0].message.content
print(results)
```

```python
# OK let's turn this into results!

results_dict = json.loads(results)
ranks = results_dict["results"]
for index, result in enumerate(ranks):
    competitor = competitors[int(result)-1]
    print(f"Rank {index+1}: {competitor}")
```

```python
# Check the current ranking to decide whether the revision loop can stop.
def is_llama_first(curr_ranks):
    return competitors[int(curr_ranks[0])-1] == 'llama3.2:latest'
```

```python
# Ask the judge to rank the supplied answers and return the ordered competitor numbers.
def get_ranks(judge_model_name, rank_message):
    messages = [{"role": "user", "content": rank_message}]
    response = openai.chat.completions.create(model=judge_model_name, messages=messages)
    results = response.choices[0].message.content

    # Parse the judge's JSON so the loop can inspect the latest ranking.
    results_dict = json.loads(results)
    print(results_dict["results"])
    return results_dict["results"]
```

```python
# Ask the evaluator to compare Llama with its peers and explain specific improvement opportunities.
def get_feedback_message(question, prev_answer, peer_answers):
    return f"""You are an evaluator helping improve one answer in a competition.

Question:
{question}

Llama's current answer:
{prev_answer}

Other models' answers:
{peer_answers}

Use the same criteria as the ranking step: clarity and strength of argument. Compare Llama's answer with the strongest peer answer(s). Explain the specific differences that make the peer answer stronger and give Llama concise, actionable suggestions to address those gaps. Do not write a replacement answer and do not favor Llama; evaluate the content fairly."""
```

```python
# Give Llama the original task, its last attempt, and the evaluator's critique for revision.
def revised_answer_prompt(question, prev_answer, feedback):
    return f"""Answer this question, using the evaluator's feedback to improve your previous answer.

Question:
{question}

Your previous answer:
{prev_answer}

Evaluator feedback:
{feedback}

Write a revised answer that addresses the useful feedback while still answering the question accurately. Make sure evaluate each response for clarity and strength of argument and improve in these category, as requested. Return only the revised answer."""
```

```python
# Format current answers with competitor numbers so the judge's ranking maps back to the lists.
def make_llm_answers():
    together = ""
    for index, answer in enumerate(answers):
        together += f"# Response from competitor {index+1}\n\n"
        together += answer + "\n\n"

    return together
```

```python
# Build a fresh judging prompt from the current answer list on every round.
def get_rank_message(question, llms_answers):
    return f"""You are judging a competition between {len(competitors)} competitors.
Each model has been given this question:

{question}

Your job is to evaluate each response for clarity and strength of argument. Judge only the answer content; do not favor any model. Return every competitor number exactly once.

Respond with JSON only, in this format:
{{"results": ["best competitor number", "second best competitor number", "third best competitor number", ...]}}

Responses:
{llms_answers}

Return the ranked order of competitor numbers. Do not include markdown or code fences."""
```

```python
# Configure the retry cap, the two models, and the ranking from the initial judging pass.
max_revision_rounds = 15
curr_run = 0
model_name = "llama3.2:latest"
judge_model_name = "gpt-5.4-nano"
curr_ranks = ranks

# Continue only while the cap is not reached and Llama is not ranked first.
while curr_run < max_revision_rounds and not is_llama_first(curr_ranks):
    # Temporarily remove Llama so the remaining entries are the fixed peer answers.
    removed_index = competitors.index(model_name)
    competitors.pop(removed_index)
    last_answer = answers.pop(removed_index)

    # Give the evaluator the peer answers so its critique can identify relative strengths.
    peer_answers = "\n\n".join(
        f"{peer_name}:\n{peer_answer}"
        for peer_name, peer_answer in zip(competitors, answers)
    )
    feedback_message = [{
        "role": "user",
        "content": get_feedback_message(question, last_answer, peer_answers),
    }]
    feedback_response = openai.chat.completions.create(
        model=judge_model_name,
        messages=feedback_message,
    )
    feedback_answer = feedback_response.choices[0].message.content

    # Ask Llama to revise its last answer using the evaluator's feedback.
    revised_answer_message = [{
        "role": "user",
        "content": revised_answer_prompt(question, last_answer, feedback_answer),
    }]
    revised_answer_response = ollama.chat.completions.create(
        model=model_name,
        messages=revised_answer_message,
    )
    revised_answer = revised_answer_response.choices[0].message.content

    # Reinsert Llama's revision, rebuild the ranking prompt, and capture the new ranks.
    record(model_name, revised_answer)
    get_ranks_messages = get_rank_message(question, make_llm_answers())
    curr_ranks = get_ranks(judge_model_name, get_ranks_messages)
    curr_run += 1

    print(f"Revision {curr_run}/{max_revision_rounds}; rank order: {curr_ranks}")

# Display the final Llama answer after reaching rank 1 or exhausting the retry cap.
display(Markdown(answers[-1]))
```
